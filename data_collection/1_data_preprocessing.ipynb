```python
import pandas as pd
import numpy as np

# Generate synthetic patient data
np.random.seed(42)

n = 1000
data = {
    "age": np.random.randint(20, 90, n),
    "hospital_stay_days": np.random.randint(1, 15, n),
    "previous_admissions": np.random.randint(0, 5, n),
    "chronic_disease": np.random.choice(["yes", "no"], n),
    "treatment_success": np.random.choice(["yes", "no"], n),
    "gender": np.random.choice(["male", "female"], n),
    "readmitted": np.random.choice([0, 1], n, p=[0.7, 0.3])  # 0=no, 1=yes
}

df = pd.DataFrame(data)
df.head()
```
